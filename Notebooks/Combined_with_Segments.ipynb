```python
import pandas as pd
from google.colab import files

uploaded = files.upload()
df = pd.read_csv('combined_with_themes.csv')
```

```python
management_technical_titles = ['Operations Manager', 'Area Manager', 'Process Assistant',
    'Pathways Operations Manager', 'Area Manager Intern', 'Sr Program Manager',
    'IT Support Engineer', 'Software Development Engineer (SDE)', 'HR Partner Intern',
    'HRA', 'Human Resource Associate Partner', 'Project Management Intern',
    'Whs Specialist', 'Quality Control', 'Técnico de mantenimiento industrial',
    'Reliability & Maintenance Engineering', 'Driver', 'Driver Associate',
    'Production', 'Seller']

def assign_segment(row):
    status = row['employee_status']
    title = row['employee_job_title']

    if pd.isnull(status):
        return 'Other'
    if status in ['CONTRACT', 'FREELANCE', 'INTERN']:
        return 'Contract & Temporary Workers'
    elif title in management_technical_titles:
        return 'Management & Specialized Roles'
    elif status == 'REGULAR':
        return 'Full-Time Warehouse Associates'
    elif status == 'PART_TIME':
        return 'Part-Time Warehouse Associates'
    else:
        return 'Other'

df['Segment'] = df.apply(assign_segment, axis=1)

print(df['Segment'].value_counts())
```

```python
df.to_csv('combined_with_segments.csv', index=False)
```
